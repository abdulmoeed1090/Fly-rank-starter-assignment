## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The dataset does not contain a column that directly tells us whether a page needs refreshing.

Therefore, I will use a **proxy target** instead of a true label.

The proxy target represents the refresh priority of a page based on observed signals such as:

- Content age
- Search impressions
- Click-through rate (CTR)
- Engagement rate
- Search trends
- Freshness metrics

These signals can be combined into a refresh score that estimates which pages are likely to benefit from an update.

This target is based on observed data rather than manually labeled outcomes.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

url = "https://raw.githubusercontent.com/abdulmoeed1090/Fly-rank-starter-assignment/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

# Example proxy score (for illustration only)
df["refresh_proxy"] = (
    df["content_age_days"] * 0.4 +
    (1 - df["ctr"]) * 100 * 0.3 +
    (1 - df["engagement_rate"]) * 100 * 0.3
)

df[["content_age_days", "ctr", "engagement_rate", "refresh_proxy"]].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as:

"If CTR is below 2%, refresh the page"

is too simple.

A page may have a low CTR but very little search volume, making it a low priority.

Another page may have a moderate CTR but extremely high impressions, making it much more valuable to improve.

The refresh decision depends on many interacting factors including:

- Search volume
- Impressions
- Clicks
- CTR
- Engagement rate
- Content age
- Freshness
- Search trends

Machine learning can learn complex relationships between these signals and produce a better ranking than a single rule.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

#This is an example fixed rule

rule_based = df[df["ctr"] < 0.02]

print("Pages selected by simple CTR rule:", len(rule_based))

print("\nAverage Search Volume of these pages:")

print(rule_based["search_volume"].mean())
```
