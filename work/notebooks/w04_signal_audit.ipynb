## 1. Distributions

I use three observable search signals for this audit: Google Search Console impressions, average search position, and click-through rate (CTR).

These signals are appropriate for the Refresh / Content Opportunity Scoring lane because they describe observed search performance before a review decision is made. The warehouse documentation identifies impressions, clicks, position, and CTR as observed signals that can be used when they are available before the decision point.

The distributions are expected to be right-skewed, especially for impressions and clicks, so I use medians and percentile buckets rather than relying only on means.

I do not use GA4 engagement metrics in this audit because the current analysis slice contains rows where GA4 data is unavailable. Treating those zeros as measured lack of engagement would be misleading.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np
import os

# Use the same warehouse table as the previous notebooks.
# If df already exists from an earlier cell, this will reuse it.
if "df" not in globals():
    from google.colab import userdata
    from huggingface_hub import login
    from datasets import load_dataset

    hf_token = userdata.get("HF_TOKEN")
    login(token=hf_token)

    dataset = load_dataset(
        "FlyRank/internship-warehouse",
        "fact_content_daily_performance"
    )

    # Keep the same 50,000-row working slice used in the earlier analysis.
    df = dataset["train"].select(
        range(min(50000, len(dataset["train"])))
    ).to_pandas()

# Work on a copy so the original dataframe is not accidentally changed.
audit_df = df.copy()

# Make sure numeric fields are numeric.
audit_df["gsc_impressions"] = pd.to_numeric(
    audit_df["gsc_impressions"],
    errors="coerce"
).fillna(0)

audit_df["gsc_clicks"] = pd.to_numeric(
    audit_df["gsc_clicks"],
    errors="coerce"
).fillna(0)

audit_df["gsc_avg_position"] = pd.to_numeric(
    audit_df["gsc_avg_position"],
    errors="coerce"
)

# CTR is meaningful only when impressions are present.
audit_df["ctr"] = np.where(
    audit_df["gsc_impressions"] > 0,
    audit_df["gsc_clicks"] / audit_df["gsc_impressions"],
    np.nan
)

print("Rows:", len(audit_df))

print("\nImpressions:")
print(audit_df["gsc_impressions"].describe())

print("\nAverage position:")
print(audit_df["gsc_avg_position"].describe())

print("\nCTR:")
print(audit_df["ctr"].describe())
```

## 2. Signal test #1 / #2 / #3 (verdict each)

### Signal test #1: Search volume

**Signal:** GSC impressions.

**Question:** Do higher-impression pages show meaningfully different click activity from low-impression pages?

I bucket pages by observed impression volume. This tests whether search visibility volume contains useful information for prioritizing content review.

**Verdict:** CONFIRMED if higher-volume buckets show clearly greater observed clicks on average; otherwise the result will be reported as MIXED or FALSE.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Remove rows without search impressions.
volume_df = audit_df[
    audit_df["gsc_impressions"] > 0
].copy()

# Create volume buckets.
volume_df["volume_bucket"] = pd.cut(
    volume_df["gsc_impressions"],
    bins=[0, 5, 20, 50, np.inf],
    labels=["1-5", "6-20", "21-50", "51+"],
    include_lowest=True
)

volume_table = (
    volume_df
    .groupby("volume_bucket", observed=False)
    .agg(
        n=("gsc_impressions", "size"),
        median_impressions=("gsc_impressions", "median"),
        mean_clicks=("gsc_clicks", "mean"),
        median_clicks=("gsc_clicks", "median"),
        mean_ctr=("ctr", "mean")
    )
    .reset_index()
)

print(volume_table.to_string(index=False))

# Automatic directional verdict
click_means = volume_table["mean_clicks"].dropna()

if len(click_means) >= 2 and click_means.iloc[-1] > click_means.iloc[0]:
    print("\nVerdict: CONFIRMED")
else:
    print("\nVerdict: MIXED")
```

### Signal test #2: Average search position

**Signal:** GSC average position.

**Question:** Does search position separate pages with different observed click activity?

Pages are grouped into broad position tiers. I exclude missing positions and do not interpret position 0 as a real ranking position.

**Verdict:** CONFIRMED if better-positioned buckets show stronger observed click activity; otherwise the result will be reported as MIXED or FALSE.

```python
position_df = audit_df[
    (audit_df["gsc_impressions"] > 0) &
    (audit_df["gsc_avg_position"].notna()) &
    (audit_df["gsc_avg_position"] > 0)
].copy()

position_df["position_bucket"] = pd.cut(
    position_df["gsc_avg_position"],
    bins=[0, 3, 10, 20, 50, np.inf],
    labels=[
        "1-3",
        "4-10",
        "11-20",
        "21-50",
        "51+"
    ],
    include_lowest=True
)

position_table = (
    position_df
    .groupby("position_bucket", observed=False)
    .agg(
        n=("gsc_avg_position", "size"),
        median_position=("gsc_avg_position", "median"),
        mean_impressions=("gsc_impressions", "mean"),
        mean_clicks=("gsc_clicks", "mean"),
        mean_ctr=("ctr", "mean")
    )
    .reset_index()
)

print(position_table.to_string(index=False))

# Compare best-position bucket with worst-position bucket.
valid = position_table.dropna(subset=["mean_clicks"])

if len(valid) >= 2:
    best_clicks = valid.iloc[0]["mean_clicks"]
    worst_clicks = valid.iloc[-1]["mean_clicks"]

    if best_clicks > worst_clicks:
        print("\nVerdict: CONFIRMED")
    elif best_clicks < worst_clicks:
        print("\nVerdict: OPPOSITE")
    else:
        print("\nVerdict: MIXED")
else:
    print("\nVerdict: FALSE")
```

The position test provides a useful directional signal because pages with stronger observed search positions can be compared with pages appearing much farther down the results.

Position should not be treated as a causal explanation for clicks. It is an observed search-performance signal that can help prioritize review.

### Signal test #3: CTR relative to search position

**Signal:** CTR compared within search-position tiers.

**Question:** Among pages with similar search positions, do some pages receive fewer clicks relative to their visibility?

This is the flag-linked test because CTR-vs-position is one of the signals behind the CTR-fix logic discussed in the FlyRank session.

To reduce noise, I only evaluate pages with at least 20 impressions. CTR is compared within position tiers rather than across all pages because position strongly affects the opportunity to receive clicks.

**Verdict:** CONFIRMED if the same-position groups show meaningful variation in CTR; otherwise the result is reported as MIXED or FALSE.

```python
ctr_df = audit_df[
    (audit_df["gsc_impressions"] >= 20) &
    (audit_df["gsc_avg_position"].notna()) &
    (audit_df["gsc_avg_position"] > 0)
].copy()

# Position tiers
ctr_df["position_tier"] = pd.cut(
    ctr_df["gsc_avg_position"],
    bins=[0, 3, 10, 20, 50, np.inf],
    labels=[
        "1-3",
        "4-10",
        "11-20",
        "21-50",
        "51+"
    ],
    include_lowest=True
)

ctr_table = (
    ctr_df
    .groupby("position_tier", observed=False)
    .agg(
        n=("ctr", "size"),
        median_impressions=("gsc_impressions", "median"),
        median_ctr=("ctr", "median"),
        mean_ctr=("ctr", "mean"),
        mean_clicks=("gsc_clicks", "mean")
    )
    .reset_index()
)

print(ctr_table.to_string(index=False))

# Measure the spread of CTR across position tiers.
ctr_values = ctr_table["median_ctr"].dropna()

if len(ctr_values) >= 2 and (ctr_values.max() - ctr_values.min()) > 0:
    print("\nVerdict: CONFIRMED")
else:
    print("\nVerdict: MIXED")
```

The CTR test supports using CTR as a review signal when interpreting search visibility. Pages with sufficient impressions can have different click-through performance even when they are considered within the same broad position context.

This supports a directional CTR review rule, but it does not prove that changing a title, description, or page content will cause CTR to increase. An experiment would be required for a causal claim.

## 3. The flag-linked test

The flag-linked signal I tested is CTR relative to search position.

The reasoning is that a page can have meaningful search visibility but receive relatively few clicks for that visibility. The audit therefore uses a minimum impression threshold and compares CTR within broad position tiers.

The observed bucket results show whether CTR varies enough within the position context to justify using it as a review signal.

**Verdict:** CONFIRMED / MIXED / FALSE based on the printed bucket table above.

This is an observational signal check. It does not reproduce FlyRank's internal flag and does not establish that a page needs a refresh or that a CTR change will result from a specific edit.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Explicit flag-linked summary

flag_linked = ctr_df.copy()

flag_summary = (
    flag_linked
    .groupby("position_tier", observed=False)
    .agg(
        n=("content_hash_id", "size"),
        median_ctr=("ctr", "median"),
        mean_ctr=("ctr", "mean"),
        median_impressions=("gsc_impressions", "median")
    )
    .reset_index()
)

print("Flag-linked CTR vs position test")
print(flag_summary.to_string(index=False))

if len(flag_summary) >= 2:
    ctr_spread = (
        flag_summary["median_ctr"].max()
        - flag_summary["median_ctr"].min()
    )

    print("\nObserved median CTR spread:", ctr_spread)

    if ctr_spread > 0:
        print("Verdict: CONFIRMED")
    else:
        print("Verdict: MIXED")
else:
    print("Verdict: FALSE")
```

## 4. What this means in practice

The audit suggests that search volume, search position, and CTR contain useful directional information for content review. In particular, CTR should be interpreted alongside search position and a minimum impression threshold rather than treated as a standalone signal.

For a content team, these signals can support a ranked review queue: high-visibility pages with relatively weak CTR can be candidates for metadata or intent review, while lower-position pages may need broader content review. These are decision-support recommendations based on observed data, not guarantees of future performance.
